```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load RGB image (OpenCV loads as BGR)
img_bgr = cv2.imread("WSe2F1_100x.jpg")

if img_bgr is None:
    raise FileNotFoundError("Image not found")
# Convert BGR → RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convert to float
img_rgb = img_rgb.astype(np.float32)

# Extract channels
R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]

# Optical intensity
intensity = 0.299 * R + 0.587 * G + 0.114 * B
plt.imshow(intensity, cmap="gray")
plt.colorbar(label="Intensity (a.u.)")
plt.title("Optical Intensity from RGB Image")
plt.axis("off")
plt.show()

print("Mean intensity:", np.mean(intensity))
print("Max intensity:", np.max(intensity))
print("Min intensity:", np.min(intensity))

# Integrated intensity (total signal)
integrated_intensity = np.sum(intensity)
print("Integrated intensity:", integrated_intensity)

# Define ROI (x, y, width, height) u can select desired region of
#the flake where u want to caclualate thickness
x, y, w, h = 1265, 750, 10, 10

roi = intensity[y:y+h, x:x+w]

roi_mean = np.mean(roi)
roi_sum = np.sum(roi)

print("ROI mean intensity:", roi_mean)
print("ROI integrated intensity:", roi_sum)

plt.imshow(roi, cmap="gray")
plt.colorbar(label="Intensity (a.u.)")
plt.title("ROI Intensity")
plt.axis("off")
plt.show()

R_int = np.sum(R)
G_int = np.sum(G)
B_int = np.sum(B)

print("Red channel intensity:", R_int)
print("Green channel intensity:", G_int)
print("Blue channel intensity:", B_int)

#The idea is to calculate the nean intensity of the flake (If) and that of
#bare substrate (Is), their ratio would give u the contrast dependence.
```
